```python
import pandas as pd

data = {
    "Customer": ["A", "B", "A", "C", "C", "A", "B", "D"],
    "Date": [2020, 2021, 2021, 2021, 2022, 2022, 2022, 2022],
    "Spend": [30.0, 50.0, 35.0, 62.0, 70.0, 45.0, 51.50, 100.0],
}

df = pd.DataFrame(data)


def classify_growth_potential(df):
    total_distinct_years = df["Date"].nunique()
    results = []

    for customer, group in df.groupby("Customer"):
        yearly_spend = group.groupby("Date")["Spend"].sum().sort_index()
        years_bought = len(yearly_spend)

        pct_change = yearly_spend.pct_change().dropna()
        has_min_5pct_growth = len(pct_change) > 0 and (pct_change >= 0.05).all()

        if years_bought == total_distinct_years and has_min_5pct_growth:
            segment = "Loyal growth"
        elif years_bought == total_distinct_years:
            segment = "Loyal"
        elif years_bought > 1 and has_min_5pct_growth:
            segment = "Regular growth"
        elif years_bought > 1:
            segment = "Regular"
        else:
            segment = "One time"

        results.append({"Customer": customer, "Growth Potential": segment})

    return pd.DataFrame(results)


output_df = classify_growth_potential(df)
print(output_df)
```
